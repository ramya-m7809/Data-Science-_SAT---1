```python
import pandas as pd

df = pd.read_csv("Q32_machine_sensors (1).csv")

print(df.head())
print(df.info())
```

```python

df = df.drop_duplicates()


print(df.isnull().sum())


df.fillna(df.mean(numeric_only=True), inplace=True)


df['failure'] = df['failure_status'].map({
    'Normal': 0,
    'Failure': 1
})

print("Cleaning Done ✅")
```

```python

print(df['failure'].value_counts())


df['temp_range'] = pd.cut(df['temperature_c'],
                         bins=[0,50,80,120],
                         labels=['Low','Medium','High'])

print(df.groupby('temp_range')['failure'].mean())


df['load_range'] = pd.cut(df['load_percent'],
                         bins=[0,40,70,120],
                         labels=['Low','Medium','High'])

print(df.groupby('load_range')['failure'].mean())
```

```python

df['stress_score'] = (
    df['temperature_c'] * 0.3 +
    df['vibration_mm_s'] * 0.4 +
    df['load_percent'] * 0.3
)

print(df[['stress_score']].head())
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.scatter(df['temperature_c'], df['vibration_mm_s'], c=df['failure'])
plt.xlabel("Temperature")
plt.ylabel("Vibration")
plt.title("Failure Pattern")
plt.show()
```

```python
import seaborn as sns

plt.figure()
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
df.to_csv("cleaned_machine_data.csv", index=False)
```
